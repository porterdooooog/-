# 便利商店咖啡計價系統
這是一個簡單的互動式便利商店咖啡計價程式，支援選擇品項、大小、冷熱、加料以及計算總金額。

```python
def calculate_coffee_price():
    # 1. 定義品項與價格
    menu = {
        '1': {'name': '美式咖啡', 'price': 45},
        '2': {'name': '拿鐵咖啡', 'price': 55},
        '3': {'name': '卡布奇諾', 'price': 55},
        '4': {'name': '焦糖瑪奇朵', 'price': 65}
    }

    size_multiplier = {
        'S': {'name': '小杯', 'add': -5},
        'M': {'name': '中杯', 'add': 0},
        'L': {'name': '大杯', 'add': 15}
    }

    extra_options = {
        '1': {'name': '加一份濃縮 (Espresso)', 'price': 15},
        '2': {'name': '加榛果糖漿', 'price': 10},
        '3': {'name': '不加料', 'price': 0}
    }

    print("=== 歡迎使用便利商店咖啡計價系統 ===\n")

    # 選擇咖啡品項
    print("請選擇咖啡品項:")
    for key, val in menu.items():
        print(f"{key}. {val['name']} (${val['price']})")

    item_choice = input("請輸入品項編號 (1-4): ").strip()
    if item_choice not in menu:
        print("無效的選擇，以美式咖啡 ($45) 計算。")
        item_choice = '1'

    base_item = menu[item_choice]
    base_price = base_item['price']

    # 選擇容量大小
    print("\n請選擇容量大小 (S:小杯-$5, M:中杯, L:大杯+$15):")
    size_choice = input("請輸入大寫字母 (S/M/L): ").strip().upper()
    if size_choice not in size_multiplier:
        print("無效的選擇，以中杯計算。")
        size_choice = 'M'

    size_info = size_multiplier[size_choice]
    base_price += size_info['add']

    # 選擇冷熱
    print("\n請選擇冷熱:")
    print("1. 冰 (Ice)")
    print("2. 熱 (Hot)")
    temp_choice = input("請輸入編號 (1-2): ").strip()
    temp_str = "冰" if temp_choice == '1' else "熱"

    # 選擇加料
    print("\n請選擇加料項目:")
    for key, val in extra_options.items():
        print(f"{key}. {val['name']} (+${val['price']})")
    extra_choice = input("請輸入編號 (1-3): ").strip()
    if extra_choice not in extra_options:
        extra_choice = '3'

    extra_item = extra_options[extra_choice]
    base_price += extra_item['price']

    # 輸入數量
    try:
        quantity = int(input("\n請輸入購買數量: "))
        if quantity <= 0:
            quantity = 1
    except ValueError:
        print("輸入格式錯誤，數量以 1 杯計算。")
        quantity = 1

    total_price = base_price * quantity

    # 輸出明細
    print("\n" + "="*30)
    print("【 購買明細 】")
    print(f"品項: {temp_str}{base_item['name']} ({size_info['name']})")
    if extra_item['price'] > 0:
        print(f"加料: {extra_item['name']}")
    print(f"單杯價格: ${base_price} 元")
    print(f"數量: {quantity} 杯")
    print("-"*30)
    print(f"總計金額: ${total_price} 元")
    print("="*30)

# 執行計價程式
calculate_coffee_price()
```
